```python
eur_list = (
    pl.read_csv('PATH_TO_FILE')
    .with_columns(
        SAMPLE = pl.col('eid').cast(pl.Utf8)
    )
    .select('SAMPLE')
)
eur_list
```

```python
!dx download project-REDACTED:/processed_data/wgs/qced_maf1e-3_loftee_olink_genes.parquet -o PATH_TO_FILE

var_df = pl.scan_parquet("PATH_TO_FILE")
var_df.head().collect()
```

```python
(
    var_df
    .join(
        eur_list.lazy(),
        on='SAMPLE',
        how='semi'
    )
    .sink_parquet("PATH_TO_FILE", engine='streaming')
)
```
