# Make plots

```python
import polars as pl

DATA_DIR = "PATH_TO_FILE"
anno = pl.read_parquet(f"{DATA_DIR}/annotations_20260713.parquet")
anno
```

```python
tmp = anno[['id', 'region', "popeve", "cpt1_llr", "bayes_del", "revel_score", "clinpred_score", "am_pathogenicity", "esm1v", "cadd_raw"]].drop_nulls().drop_nans()
tmp = tmp.unpivot(
    index=['id', 'region'],
    variable_name='annotation',
    value_name='score'
)
tmp
```

```python
from plotnine import *

(
    ggplot(tmp, aes(x="score"))
    + geom_histogram(bins=50, fill='steelblue', color='black', alpha=0.7)
    + scale_y_log10()
    + facet_wrap('annotation', scales='free', ncol=1)
    + theme_minimal()
    + theme(
        axis_text_x=element_text(rotation=45, hjust=1),
        figure_size=(5, 12)
    )
)
```
